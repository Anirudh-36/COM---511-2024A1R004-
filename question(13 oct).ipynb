```python
"""
2.  Instance methods : Rectangle area and perimeter
Problem : Create a Rectangle class with length and breadth. Use instance methods to calculate its area and perimeter.
"""
class Rectangle :
    def __init__(self,length,breadth):
        self.length = length
        self.breadth = breadth
    def area(self):
        return self.length * self.breadth
    def perimeter(self):
        return 2 * (self.length + self.breadth)
r = Rectangle(12,13)
print("Area :",r.area())
print("Perimeter :",r.perimeter())
```

```python
"""
3. Create a Product class with the product name, price and quantity.Add an instance method to calculate the total cost.Take these values from the user and 
display the product name and total cost.
"""
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def total_cost(self):
        return self.price * self.quantity
    
name = input("Enter product name: ")
price = float(input("Enter price: "))
quantity = int(input("Enter quantity: "))

p = Product(name, price, quantity)

print("Product Name:", p.name)
print("Total Cost:", p.total_cost())
```

```python
"""
4. Create a BankAccount class that stores an account holder's name and balance.Add instance methods to deposit money,
withdraw money, and display the balance.Do not allow a withdraw greater than the available balance.
"""
class BankAccout :
    def __init__(self,name,balance):
        self.name = name
        self.balance = balance
    def deposit(self,amount):
        self.balance += amount
    def withdraw(self,amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            print("Insufficient Balance")
    def display(self):
        print("Account Holder :",self.name)
        print("Balance :",self.balance)
name = input("Enter Name :")
balance = float(input("Enter Balance :"))

b = BankAccout(name,balance)

deposit = float(input("Enter Deposit amount :"))
b.deposit(deposit)

withdraw = float(input("Enter Withdraw amount :"))
b.withdraw(withdraw)

b.display()
```

```python
"""
6. Create an ElectricityBill class that stores a customer's name and electricity units used.Calculate the bill using three rates: rupees 3 per unit
for the first 100 units, rupees 5 per unit for the next 100 units, and rupees 7 per unit for units above 200. Add a fixed charge of rupees 50.Display the 
customer's name,units used, and total bill. If the entered units are negative,display an error messsage.    
"""
class ElectricityBill:
    def __init__(self, name, units):
        self.name = name
        self.units = units

    def calculate(self):
        if self.units < 0:
            print("Invalid units")
        else:
            if self.units <= 100:
                bill = self.units * 3
            elif self.units <= 200:
                bill = 100 * 3 + (self.units - 100) * 5
            else:
                bill = 100 * 3 + 100 * 5 + (self.units - 200) * 7

            bill = bill + 50

            print("Customer Name:", self.name)
            print("Units Used:", self.units)
            print("Total Bill: Rs.", bill)


name = input("Enter customer name: ")
units = int(input("Enter units used: "))

obj = ElectricityBill(name, units)
obj.calculate()
```

```python
"""
9. Write a Python program that creates a ParkingTicket class for calculating the parking fee for one vehicle.
The program must :
1. Ask the user to enter the vehicle number and parking duration in hours. The duration may include a fraction, such as 2.5 hours.
2. Calculate the fee using the billable hours:
- Each started hour counts as one full hour. For example, 2.1 hours is billed as 3 hours.
- Charge rupees 30 per billable hour for the first 20 hours.
- Charge rupees 20 for each billable hour after the first two hours.
3. Display the vehicle number, actual parking duration, and total fee.
4. If the duration is zero or negative, display an error message and do not calculate a fee. 
"""
import math

class ParkingTicket:
    def __init__(self, number, hours):
        self.number = number
        self.hours = hours

    def calculate(self):
        if self.hours <= 0:
            print("Invalid parking duration")
        else:
            bill_hours = math.ceil(self.hours)

            if bill_hours <= 2:
                fee = bill_hours * 30
            else:
                fee = 2 * 30 + (bill_hours - 2) * 20

            print("Vehicle Number:", self.number)
            print("Parking Duration:", self.hours, "hours")
            print("Total Fee: Rs.", fee)


number = input("Enter vehicle number: ")
hours = float(input("Enter parking duration in hours: "))

obj = ParkingTicket(number, hours)
obj.calculate()
```
