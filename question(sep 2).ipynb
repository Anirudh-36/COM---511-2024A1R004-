```python
# 7. Write a Python program to take a sentence and replace all spaces with underscores.
sentence = input("Enter a sentence :")
sentence1 = sentence.replace(" ", "_")
print("Updated sentence :", sentence1)
```

```python
# 8. Write a Python program to take a sentence, detect double spaces, and replace then with single spaces.
sentence = input("Enter a sentence :")
sentence1 = sentence.replace("  "," ")
print("Updated sentence :",sentence1)
```

```python
# 9. Write a Python program to take a word and print it in reverse order using slicing.Also check whether it is the same forward and backward.
word = input("Enter a word :")
reverse = word[::-1]
print("Reverse = ",reverse)
print("Palindrome:", word == reverse)
```

```python
# 11. Write a Python program to take a word and print every second character.
word = input("Enter a word :")
result = word[1::2]
print("Every Second character :",result)
```

```python
# 12. Write a Python program to take a password and check whether it contains @ and has at least 8 characters.
password = input("Enter a Password :")
if "@" in password and len(password) >= 8 :
    print("Valid Password")
else:
    print("Invalid Password")
```

```python
# 13. Write a Python program to take a string and seperate characters present at even index positions and odd index positions.
string = input("Enter a String :")

even = string[::2]
odd = string[1::2]

print("Even index characters:", even)
print("Odd index characters:", odd)
```

```python
# 14. Take an email address and check whether it contains @ and .com.
email = input("Enter email address: ")
print("Valid Email:", "@" in email and ".com" in email)
```
