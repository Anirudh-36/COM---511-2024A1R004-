```python
# 7. Write a program to take temperature in Celsius from the user and convert it into Fahrenheit
celsius = float(input("Enter temperature in Celsius: "))
fahrenheit = (celsius * 9/5) + 32
print("Temperature in Fahrenheit =", fahrenheit)
```

```python
# 8. Write a python program to calculate simple Interest and total amount using Principal,Rate, and Time entered by the user. 
P = float(input("Enter Principal : "))
R = float(input("Enter Rate : "))
T = float(input("Enter Time : "))

SI = (P * R * T) / 100
Amount = P + SI

print("Simple Interest =", SI)
print("Total Amount =", Amount)
```

```python
# 9. Write a Python program to take distance in Kilometers and convert it into meters,centimeters, and millimeters.
km = float(input("Enter distance in Kilometers: "))
meters = km * 1000
centimeters = km * 100000
millimeters = km * 1000000

print("Meters =", meters)
print("Centimeters =", centimeters)
print("Millimeters =", millimeters)
```

```python
# 10. Write a Python program to ask the user for radius and calculate area and circumference of circle.
radius = float(input("Enter Radius :"))
area = 3.14 * radius * radius
circumference = 2 * 3.14 * radius
print("Area = ",area)
print(f"Circumference :{circumference:.2f}")
```
