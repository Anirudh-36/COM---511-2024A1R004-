```python
# 1. Write a Python program to store two points as tuples and calculate the distance between them.
import math

x1 = int(input("Enter x1 :"))
y1 = int(input("Enter y1 :"))

x2 = int(input("Enter x2 :"))
y2 = int(input("Enter y2 :"))

point1 = (x1, y1)
point2 = (x2, y2)

distance = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)

print("Point 1 :", point1)
print("Point 2 :", point2)
print("Distance :", distance)
```

```python
# 2. Write a Python program to store all month names in a tuple.Input a month number and display the corresponding month name.
months = ("January", "February", "March", "April", "May", "June",
          "July", "August", "September", "October", "November", "December")

month = int(input("Enter month number :"))

if month >= 1 and month <= 12:
    print("Month :", months[month - 1])
else:
    print("Invalid month number")
```

```python
# 3. Write a Python program to show that tuple values cannot be changed directly.Convert tuple into list,update it, and convert it back into tuple.
t = (10, 20, 30)
print("Original tuple :", t)

l = list(t)
l[1] = 50

t = tuple(l)
print("Updated tuple :", t)
```

```python
# 4. Write a Python program to store repeated values in a tuple and count how many times a given value appears.
t = (10, 20, 10, 30, 10, 20)
value = int(input("Enter value :"))
count = t.count(value)
print("Value", value, "appears", count, "times")
```

```python
# 5. Write a Pyton program to check whether a given value is present in a tuple. if present, display its position.
t = (10, 20, 30, 40, 50)
value = int(input("Enter value :"))

if value in t:
    print("Value is present")
    print("Position :", t.index(value))
else:
    print("Value is not present")
```

```python
# 6. Write a Python program to store one student data in a tuple : name, roll number, and marks. Display grade based on marks.
name = input("Enter name :")
roll = int(input("Enter roll number :"))
marks = int(input("Enter marks :"))

student = (name, roll, marks)

if marks >= 90:
    grade = "A"
elif marks >= 80:
    grade = "B"
elif marks >= 70:
    grade = "C"
elif marks >= 60:
    grade = "D"
else:
    grade = "F"

print("Student :", student)
print("Grade :", grade)
```

```python
# 7. Write a Python program to store multiple student records as a list of tuples. Each tuple should contain name, roll number, and marks. Display Students who scored about 75.
students = [
    ("Anirudh", 1, 80),
    ("Keshav", 2, 65),
    ("Manyia", 3, 90),
    ("Akash", 4, 70)
]

for student in students:
    if student[2] > 75:
        print(student)
```
