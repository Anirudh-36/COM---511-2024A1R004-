2. **Memory Management:** Automatic.

# How Python Works

Python is commonly called an **interpreted language**.

**CPython** first compiles valid source code into **bytecode**, and then the bytecode is executed by the Python Virtual Machine (PVM).

# Data Types

A data type is used to specify **what type of data we need to store**.

### Primarily, these are the following data types:

- `int`
- `None`
- `float`
- `boolean`
- `double`

> **Note:** Python does not have a separate `double` data type. `float` is used for floating-point numbers.

---

# Rules for Choosing an Identifier

- A variable name can contain **alphabets, digits, and underscore (`_`)**.
- It cannot be started with a **number**.
- It **can be started with an underscore**.
